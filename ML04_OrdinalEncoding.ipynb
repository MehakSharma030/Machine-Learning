```python
import pandas as pd
```

```python
df = pd.DataFrame({
    "Size": ['s', 'xl', 'l', 'm', 's', 'xl', 'xl', 's', 's', 'm', 'l', 'l']
})
```

```python
df
```

# ordinal encoding using SCIKIT-LEARN

```python
from sklearn.preprocessing import OrdinalEncoder
```

```python
ord_data = [['s' , 'm' , 'l' , 'xl']]
```

```python
oe = OrdinalEncoder( categories = ord_data)
oe.fit( df[["Size"]] )
```

```python
df["Size_enc"] = oe.transform( df[["Size"]] )
```

```python
df
```

# ordinal encoding USING MAP CODING

```python
ord_data_1 = { 's' : 0 , 'm' : 1 , 'l' : 2 , 'xl' : 3 }
```

```python
df["Size_enc_map"] = df["Size"].map(ord_data_1)
```

```python
df
```
