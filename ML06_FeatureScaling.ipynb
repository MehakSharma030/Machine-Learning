```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv("loan.csv")
```

# **STANDARDIZATION**

```python
data.isnull().sum()
```

```python
data.head(3)
```

```python
# for coloumn applicant income there are not null values
sns.distplot(x = data['ApplicantIncome'])
```

```python
data.describe()
```

using scikit learn for applying feature scaling

```python
from sklearn.preprocessing import StandardScaler
```

```python
se = StandardScaler()
se.fit(data[['ApplicantIncome']])
```

```python
data['ApplicantIncome__fs'] = pd.DataFrame(se.transform(data[['ApplicantIncome']]) , columns = ['x'])
```

```python
data.head(3)
```

```python
data.describe()
```

```python
plt.subplot(1,2,1)
sns.distplot(data["ApplicantIncome"])
plt.title("Before")

plt.subplot(1,2,2)
sns.distplot(data["ApplicantIncome__fs"])
plt.title("After")
```

# **NORMALISATION**

```python
data = pd.read_csv("loan.csv")
```

```python
data.head()
```

```python
data.describe()
```

```python
sns.distplot(data["CoapplicantIncome"])
```

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
ms = MinMaxScaler()
```

```python
ms.fit(data[["CoapplicantIncome"]])
```

```python
ms.transform(data[["CoapplicantIncome"]])
```

```python
data["CoapplicantIncome__mms"] = ms.transform(data[["CoapplicantIncome"]])
```

```python
data.head()
```

```python
data.describe()
```

```python
plt.figure(figsize = (10,5))
plt.subplot(1,2,1)
sns.distplot(data["CoapplicantIncome"])
plt.title("Before")

plt.subplot(1,2,2)
sns.distplot(data["CoapplicantIncome__mms"])
plt.title("After")
```
